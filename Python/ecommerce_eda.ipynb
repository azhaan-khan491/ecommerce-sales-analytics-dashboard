```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
df = pd.read_csv("D:\Ecommerce_Sales_Analysis\Dataset\Superstore Cleaned 1.csv")

print(df.shape)
df.head()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Ship_Date'] = pd.to_datetime(df['Ship_Date'])
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

 # visual 1: Sales by Category

```python
Sales_by_category = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x= Sales_by_category.index,  y= Sales_by_category.values,  palette="viridis")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

 # Visual 2: Profit by Category

```python
Profit_by_Category = (df.groupby("Category")["Profit"].sum().sort_values(ascending=False))
plt.figure(figsize=(8,5))
sns.barplot (x = Profit_by_Category.index,  y = Profit_by_Category.values ,  palette ="magma")

plt.title("Profit_by_Category")
plt.xlabel("Category")
plt.ylabel("profit")
plt.show()
```

# Visual 3: Sales by Region

```python
Region_Sales = (df.groupby)("Region")["Sales"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
sns.barplot (x= Region_Sales.index , y= Region_Sales.values, palette="Blues")

plt.title("Sales By Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

# Visual 4: Segment Contribution

```python
Segment_Sales = (df.groupby)("Segment")["Sales"].sum()
plt.figure(figsize=(7,7))
plt.pie (x= Segment_Sales, labels= Segment_Sales.index, autopct= '%1.1f%%')

plt.title("Sales Contribution by Segment")
plt.show()
```

# Visual 5: Top 10 States by Sales

```python
Top_States = (df.groupby)("State")["Sales"].sum().sort_values(ascending=False).head(10)
plt.Figure(figsize=(12,6))
sns.barplot (x=Top_States.index, y= Top_States.values, palette="crest")

plt.title("Top 10 States by Sales")
plt.xticks(rotation = 40)
plt.xlabel("States")
plt.ylabel("Sales")
plt.show()
```

# Visual 6: Monthly Sales Trend

```python
monthly_sales = (df.groupby(df["Order_Date"].dt.to_period('M'))["Sales"].sum())

plt.figure(figsize=(15,6))

monthly_sales.plot(kind='line', marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

# Visual 7: Monthly Profit Trend

```python
Monthly_Profit =(df.groupby)(df["Order_Date"].dt.to_period("M"))["Profit"].sum()
plt.figure(figsize=(15,6))

Monthly_Profit.plot(kind = "line" , color = "green", marker ="o")
plt.title("Monthly Profit Trend")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.grid()
plt.show()
```

# Visual 8: Discount vs Profit Analysis

```python
plt.figure(figsize=(10,6))

sns.scatterplot(data= df, x = "Discount" , y = "Profit")
plt.title("Discount VS Profit")
plt.show()
```

# Visual 9: Top 10 Customers by Sales

```python
Top_Customers = df.groupby("Customer_Name")["Sales"].sum().sort_values(ascending=False).head(10)

plt.Figure(figsize=(12,6))
sns.barplot (x= Top_Customers.values , y = Top_Customers.index , palette= "viridis")
plt.title("Top 10 Customer by Sales")
plt.xlabel("Sales")
plt.ylabel("Customer Name")
plt.show()
```

# Visual 10: Top 10 Products by Sales

```python
Top_Product = df.groupby("Product_Name")["Sales"].sum().sort_values(ascending=False).head(10)

plt.Figure(figsize=(12,8))
sns.barplot (x= Top_Product.values , y = Top_Product.index , palette ="magma")
plt.title("Top 10 Product by Sales")
plt.xlabel("Sales")
plt.ylabel("Product Name")
plt.show()
```

# Visual 11: Top 10 Loss-Making Products

```python
Loss_Product = (df.groupby("Product_Name")["Profit"].sum().sort_values().head(10))

plt.Figure(figsize=(12,8))
sns.barplot (x = Loss_Product.values, y = Loss_Product.index, palette= "Reds")
plt.title("TOP 10 LOSS-MAKING PRODUCTS")
plt.xlabel("Profit")
plt.show()
```

# Visual 12: Correlation Heatmap

```python
corr = df[["Sales","Quantity","Discount","Profit"]].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap= "coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

# Visual 13: Shipping Mode Analysis

```python
Ship_Mode_Sales = (df.groupby("Ship_Mode")["Sales"].sum().sort_values(ascending=False))

plt.Figure(figsize=(8,5))

sns.barplot(x= Ship_Mode_Sales.index , y = Ship_Mode_Sales.values, palette= "Set2")
plt.title("Ship Mode Analysis")
plt.show()
```

# Visual 14: State-Level Heat Analysis

```python
Top_States = (df.groupby("State")["Sales"].sum().sort_values(ascending=False).head(15))

plt.figure(figsize=(12,8))

sns.barplot(x = Top_States.values, y = Top_States.index, palette="rocket")
plt.title("Top 15 States by Sales")
plt.show()
```
